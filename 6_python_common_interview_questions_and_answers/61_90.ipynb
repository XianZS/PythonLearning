## 65. Python 中的 Map 函数有什么作用? 怎么使用?

## 67. Python 中 reduce 函数有什么作用? 怎么使用?

res = reduce(lambda x,y: x+y, lst, 10)
print(res)  # 20，等价于 (((10+1)+2)+3)+4

## 69. 什么是 Python 的生成器?
生成器（Generator）是特殊的迭代器，惰性计算、按需生成值，极大节省内存。

两种创建方式：
1. **生成器表达式**：将列表推导式的 `[]` 换成 `()`
```python
g = (x for x in range(10))
```
2. **生成器函数**：函数中使用 `yield` 关键字，调用函数不立即执行，返回生成器对象；每次 `next()` 执行到 `yield` 处返回值并暂停，下次调用从暂停处继续。

```python
def gen(n):
    for i in range(n):
        yield i * 2

g = gen(5)
print(next(g))  # 0
print(next(g))  # 2
```

特点：惰性求值、只能遍历一次、内存占用极低，适合大数据、无限序列场景。

````python
# nums=[]
# for x in range(1,11):
#     nums.append(x)
# nums
# nums=[x for x in range(1,11)]
# nums

# --- 生成器表达式
nums1 = [x for x in range(1, 11)]
nums2 = (x for x in range(1, 11))


# next(nums2)
# next(nums2)
# --- 生成器函数
def pr(n=11):
    for x in range(1, n):
        yield x


func = pr()
# type(func)
print(f"[index=1]:{next(func)}")
print(f"[index=2]:{next(func)}")
print(f"[index=3]:{next(func)}")
````

## 71. Python 的迭代器和生成器有什么区别?
生成器是迭代器的一种，属于包含关系，核心差异：
1. **创建方式**
   - 迭代器：通过类实现 `__iter__` + `__next__` 创建，或用 `iter()` 转换可迭代对象。
   - 生成器：通过生成器函数（`yield`）或生成器表达式创建，语法更简洁。
2. **状态维护**
   - 自定义迭代器需手动维护状态、处理 `StopIteration`。
   - 生成器自动维护状态，`yield` 自动暂停/恢复，自动抛出结束异常。
3. **灵活性**：迭代器可实现复杂迭代逻辑；生成器适合简单序列生成，代码更精简。
4. **范围**：所有生成器都是迭代器，迭代器不一定是生成器。

## 73. Python 的 `__init__` 和 `__new__` 方法有什么区别?
二者都是类的构造相关魔法方法，执行顺序与作用完全不同：
1. **作用不同**
   - `__new__`：**构造方法**，负责创建并返回实例对象，是类级方法，第一个参数为 `cls`（类本身）。
   - `__init__`：**初始化方法**，负责给已创建的实例初始化属性，是实例级方法，第一个参数为 `self`（实例对象）。
2. **执行顺序**：创建对象时先执行 `__new__` 创建实例，再执行 `__init__` 初始化。
3. **返回值**：`__new__` 必须返回类的实例，否则 `__init__` 不会执行；`__init__` 无返回值（只能返回 None）。
4. **场景**：`__new__` 用于控制实例创建（单例模式、不可变类定制）；`__init__` 用于常规属性初始化。

```python
class Person:
    def __new__(cls, name):
        print("调用 __new__")
        return super().__new__(cls)
    def __init__(self, name):
        print("调用 __init__")
        self.name = name
```

````python
class People:
    def __new__(cls, name, age):
        print(f"[new]>>>调用{name}={age}")
        return super().__new__(cls)

    def __init__(self, name, age):
        print(f"[init]>>>{name}={age}")
        self.name = name
        self.age = age


p1 = People("jom", 22)
p1.name, p1.age
````

## 74. Python 函数参数 `*args` 和 `**kwargs` 有什么区别? 怎么使用?

- `**kwargs`：接收任意数量的**关键字参数**，打包成字典。`kwargs` 是约定名称，可自定义。

````python
def main(name, *args, **kwargs):
    print(name)
    print(args)
    print(kwargs)
    # pass


main("jom", "kom", "lom", "some", address="shanghai", number="123", age=22)
````

## 76. 什么是 "猴子补丁"(monkey patching) ?
猴子补丁指在**程序运行时动态修改类、模块或方法的行为**，无需修改原始源码，是动态语言的特性。

```python
class A:
    def func(self):
        print("原始方法")

def new_func(self):
    print("补丁方法")

# 运行时替换类方法
A.func = new_func
a = A()
a.func()  # 输出 "补丁方法"
```

应用场景：临时修复第三方库 Bug、测试时 Mock 替换方法、动态扩展功能。
> 注意：会增加代码隐晦度和维护成本，非必要不建议大量使用。

````python
# 已经存在功能a
# 不能满足业务需求
# a的源代码不允许你修改，因为可能有一些老业务依赖于a源代码
# 新业务只修改修改a的一部分
class People:
    def __init__(self, name):
        self.name = name

    def pr(self):
        print(f"[name] >>> {self.name}")


def pr(self):
    for x in range(10):
        print(f"[{x}-name] >>> {self.name}")


People.pr = pr
p = People("kom")
p.pr()
````

## 81. 什么是 Python 的 OOP (面向对象编程) ?
OOP（Object-Oriented Programming）即面向对象编程，是以对象为基本单元、将数据与操作数据的方法封装在一起的编程范式。

四大核心特性：
1. **封装**：绑定数据与方法，隐藏内部细节，只暴露必要接口。
2. **继承**：子类复用父类属性方法，支持扩展与重写。
3. **多态**：同一接口可对应不同实现，调用时按对象类型自动适配。
4. **抽象**：提取事物共同特征，忽略非核心细节，定义统一规范。

优势：代码复用性高、可维护性强、扩展性好，适合大型复杂项目。

## 82. 什么是 Python 面向对象的抽象特性?
抽象指隐藏事物的复杂实现细节，只对外暴露核心功能接口，使用者无需关心内部逻辑。

Python 实现抽象的方式：
1. **抽象类与抽象方法**：通过 `abc` 模块实现。抽象类不能实例化，子类必须实现所有抽象方法，否则也无法实例化。
2. 普通类封装也属于抽象，通过私有属性/方法隐藏内部细节。

```python
from abc import ABC, abstractmethod
class Animal(ABC):
    @abstractmethod
    def speak(self):
        pass  # 抽象方法，仅声明无实现

class Dog(Animal):
    def speak(self):
        print("汪汪叫")
```
作用：定义统一接口规范，强制子类遵循约定，提升代码规范性。

````python
from abc import ABC, abstractmethod


# 抽象类 --- 定义了后续实现的规范
class Animal(ABC):
    @abstractmethod
    def speak(self):
        pass


class Cat(Animal):
    def speak(self):
        print("喵喵喵")


class Dog(Animal):
    def speak(self):
        print("汪汪汪")


cat = Cat()
dog = Dog()
cat.speak(), dog.speak()
````

## 83. 什么是 Python 面向对象的封装特性?
封装指将数据与操作数据的方法绑定，隐藏内部实现与数据，仅通过公开接口访问，保障数据安全与易用性。

Python 封装层级：
1. **公有成员**：普通命名，类内外均可直接访问。
2. **受保护成员**：单下划线开头（如 `_name`），约定类内部和子类可访问，外部不建议访问。
3. **私有成员**：双下划线开头（如 `__age`），Python 会做名称改写（`_类名__属性名`），外部无法直接访问，只能通过类内部方法访问。

```python
class Person:
    def __init__(self, name, age):
        self.name = name        # 公有
        self._age = age         # 受保护
        self.__password = 123   # 私有
    def get_password(self):
        return self.__password  # 公开接口访问私有属性
```

````python
class Stu:
    def __init__(self, name, age, address):
        # 体现出公有成员的访问规则
        self.name = name
        # 体现出受保护成员的访问规则
        self._age = age
        # 私有成员的访问规则
        self.__address = address


class Child(Stu):
    def __init__(self, name, age, address):
        super().__init__(name, age, address)

    def pr(self):
        print(f"[age] >>> {self._age}")

    def pr_address(self):
        print(f"[address] >>> {self.__address}")


# stu=Stu("kom",22,"xian")
# stu.name

# child=Child("kom",22,"Xi'an")
# child.pr()
# child._age

# child=Child("kom",22,"Xi'an")
# child.pr_address()
````

## 84. 什么是 Python 面向对象的多态特性?
多态指“一个接口，多种实现”：同一个方法名，不同对象调用会产生不同执行结果。

Python 是动态语言，天然支持多态（鸭子类型）：只要对象拥有对应方法即可调用，不严格限制类型。

```python
class Dog:
    def speak(self):

    def speak(self):

```

优势：提升代码灵活性与扩展性，减少分支判断，符合开闭原则。

## 85. Python 是否支持多重继承?
Python 支持多重继承，即一个子类可同时继承多个父类，拥有所有父类的属性与方法。

```python
class A: pass
class B: pass
class C(A, B):  # 同时继承 A 和 B
    pass
```

核心问题：**方法解析顺序（MRO）**。多个父类有同名方法时，Python 通过 C3 算法确定调用优先级，可通过 `类名.__mro__` 查看。

> 注意：多重继承会增加代码复杂度，易出现命名冲突，使用需谨慎，优先考虑组合模式。

````python
class A:
    pass


class B(A):
    pass


class C(A):
    pass


class D(B, C):
    pass


print(f"[B MRO] >>> {B.__mro__}")
print(f"[C MRO] >>> {C.__mro__}")
print(f"[D MRO] >>> {D.__mro__}")
````

## 86. 为什么 Python 执行速度慢，如何改进?
速度慢的核心原因：
1. 解释型语言，逐行解释执行，无编译优化。
2. CPython 存在 GIL 全局解释器锁，多线程无法利用多核并行计算。
3. 动态类型，运行时频繁检查类型，无法做编译期优化。
4. 一切皆对象，每个数据都有额外元数据开销。

提速方案：
1. 优先使用内置函数与数据结构（底层 C 实现）。
2. CPU 密集型用多进程，绕过 GIL 利用多核。
3. 核心逻辑用 C 扩展（Cython、ctypes）。
4. 使用 PyPy 解释器，JIT 即时编译加速。
5. 数值计算用 NumPy、Pandas 等向量化库。
6. IO 密集型用 asyncio 异步编程，减少等待损耗。

## 87. 如何分析 Python 代码的执行性能?
常用性能分析工具，从粗到精定位瓶颈：
1. **time / timeit**：简单计时，测试单段代码执行耗时。
2. **cProfile**：官方内置工具，统计每个函数的调用次数、总耗时、单次耗时，定位最慢函数。
   用法：`python -m cProfile -s cumulative test.py`
3. **line_profiler**：逐行统计代码执行时间，定位具体慢的代码行。
4. **memory_profiler**：分析内存占用，定位内存泄漏与高内存点。
5. **py-spy**：采样型分析工具，无需修改代码，可附加到运行进程，适合线上环境。

分析思路：先用 cProfile 定位慢函数，再用 line_profiler 定位具体行，针对性优化。

## 88. Python 中如何读取大文件，例如内存只有 4G，如何读取一个大小为 8G 的文件
核心思路：**分块读取、逐行处理**，避免一次性加载全部内容。

常用方案：
1. **逐行迭代读取（最常用）**
文件对象本身是可迭代对象，for 循环逐行读取，每次仅加载一行。
```python
with open('big_file.txt', 'r', encoding='utf-8') as f:
    for line in f:
        process(line)  # 处理每一行
```
适合按行处理的文本文件，内存占用极低。

2. **指定字节分块读取**
适合无换行符的二进制文件或大文件，每次读取固定大小。
```python
chunk_size = 1024 * 1024  # 每次读 1MB
with open('big_file', 'rb') as f:
    while True:
        chunk = f.read(chunk_size)
        if not chunk:
            break
        process(chunk)
```

> 注意：避免使用 `read()`、`readlines()`，处理完的数据及时释放，不要全部存入列表。

## 91. Python 中如何实现多线程?
主要通过 `threading` 模块实现，两种常用方式：
1. **函数式创建**
```python
import threading
def task(name):
    print(f"线程{name}执行")

t1 = threading.Thread(target=task, args=('A',))
t2 = threading.Thread(target=task, args=('B',))
t1.start()
t2.start()
t1.join()  # 等待线程结束
t2.join()
```

2. **继承 Thread 类**
```python
class MyThread(threading.Thread):
    def __init__(self, name):
        super().__init__()
        self.name = name
    def run(self):  # 重写 run 方法
        print(f"线程{self.name}执行")

t = MyThread('C')
t.start()
```

> 注意：CPython 受 GIL 限制，多线程不适合 CPU 密集型任务，适合 IO 密集型场景；多线程共享全局变量需加锁防止数据竞争。

## 92. 请介绍 Python 中多线程和多进程的应用场景，以及优缺点?
### 多线程
- 原理：同一进程内的多个线程，共享进程内存空间，切换开销小。
- 优点：创建/切换开销低、资源占用少、线程间通信方便。
- 缺点：CPython 受 GIL 限制，CPU 密集型无法并行加速；共享资源需加锁，易引发死锁。
- 适用场景：网络爬虫、文件读写、数据库操作等 IO 密集型任务。

### 多进程
- 原理：每个进程有独立内存空间与 GIL，真正实现多核并行。
- 优点：可利用多核 CPU、进程间独立稳定性高、适合 CPU 密集型。
- 缺点：进程创建/切换开销大、进程间通信复杂。
- 适用场景：数值计算、图像处理、大数据处理等 CPU 密集型任务。

总结：IO 密集型用多线程，CPU 密集型用多进程。

## 93. 请解释 Python 线程池的工作原理?
线程池是线程复用机制：预先创建一定数量的线程放入池中，任务提交后分配给空闲线程执行，执行完毕线程不销毁，继续等待下一个任务。

工作流程：
1. 初始化：创建指定数量的工作线程，同时维护一个任务队列。
2. 提交任务：用户将任务提交到任务队列。
3. 调度执行：空闲线程从队列取任务执行。
4. 线程复用：任务执行完线程不退出，回到池中等待新任务。
5. 关闭回收：所有任务结束后关闭线程池，回收资源。

Python 主要通过 `concurrent.futures.ThreadPoolExecutor` 实现：
```python
from concurrent.futures import ThreadPoolExecutor
def task(n):
    return n * n

with ThreadPoolExecutor(max_workers=5) as executor:
    futures = [executor.submit(task, i) for i in range(10)]
    for f in futures:
        print(f.result())
```

优势：避免频繁创建销毁线程的开销、控制线程数量防止资源耗尽、简化多线程编码。
