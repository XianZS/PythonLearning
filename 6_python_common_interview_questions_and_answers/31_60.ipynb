## 31. 如何在Python中管理内存？
`Python`由解释器自动管理内存，核心机制：

1. **引用计数**：每个对象维护引用计数，计数为0时释放内存；引用增加/减少对应赋值、销毁等操作。
2. **垃圾回收**：通过标记-清除解决循环引用问题；分代回收（0/1/2三代）优化回收效率，存活越久的对象回收频率越低。
3. **内存池**：小内存（<256KB）由内存池管理，避免频繁向系统申请释放；大内存直接由系统分配。

开发者层面：避免循环引用、及时释放无用引用、使用生成器减少内存占用、通过`gc`模块调试内存问题。

## 33. Python程序退出时，是否释放所有内存分配？
从**操作系统层面**：进程结束后，系统会回收进程全部内存。
从**Python解释器层面**：不会将所有内存归还给操作系统：
1. 循环引用且关联全局变量的对象，可能无法及时回收。
2. C扩展申请的内存，需扩展自身管理，Python解释器不负责释放。
3. Python内存池中的内存会保留复用，不会立即归还系统。

## 54. Python 中 read、readline、readlines 有哪些区别?
三者都是文件对象的读取方法，核心差异在读取粒度、返回值和内存特性：
- `read(size=-1)`：一次性读取文件全部内容，返回字符串；传入`size`时读取指定字节数。适合小文件，大文件使用会占满内存。
- `readline(size=-1)`：逐行读取，每次读取一行并返回字符串；`size`可限制单行读取的字符数。内存占用低，适合逐行处理大文件。
- `readlines(hint=-1)`：一次性读取所有行，返回由每行字符串组成的列表；`hint`可限制读取总字符数。同样不适合超大文件，会一次性加载全部内容。

```python
with open('test.txt', 'r', encoding='utf-8') as f:
    content = f.read()      # 读取全部
    line = f.readline()     # 读取一行
    lines = f.readlines()   # 读取所有行成列表
```
> 补充：遍历大文件推荐直接迭代文件对象 `for line in f:`，本质逐行读取，内存效率最高。

## 55. 什么是 Python 面向对象中的继承特点?
继承是面向对象三大特性之一，指子类（派生类）复用父类（基类）的属性与方法，并可扩展自身逻辑。Python 继承的核心特点：
1. **代码复用**：子类直接继承父类属性和方法，避免重复编码。
2. **多重继承**：支持一个子类同时继承多个父类，通过 MRO（方法解析顺序）确定方法调用优先级。
3. **方法重写**：子类可覆盖父类方法，调用时优先执行子类逻辑。
4. **扩展能力**：子类可新增专属属性和方法。
5. **`super()` 调用**：可在子类中调用父类方法，常用于构造方法初始化。
6. **传递性**：C 继承 B、B 继承 A，则 C 也拥有 A 的属性与方法。

```python
class Animal:
    def __init__(self, name):
        self.name = name
    def speak(self):
        print(f"{self.name} 发出声音")

class Dog(Animal):
    def speak(self):  # 方法重写
        print(f"{self.name} 汪汪叫")
    def run(self):    # 子类新增方法
        print(f"{self.name} 奔跑")
```

## 56. Python 中深拷贝和浅拷贝有什么区别?
核心区别在于对嵌套可变对象的复制层级：
- **浅拷贝**：创建新对象，但仅复制嵌套对象的引用，内部可变对象共享内存。修改嵌套数据时，原对象与拷贝对象都会受影响。实现方式：`copy.copy()`、列表切片 `[:]`、`list()`/`dict()` 工厂方法。
- **深拷贝**：创建完全独立的新对象，递归复制所有层级的数据，嵌套对象也会重新创建。修改拷贝对象完全不影响原对象。实现方式：`copy.deepcopy()`。

> 注意：对于字符串、数字、元组等不可变对象，深浅拷贝无实质差异。

```python
import copy
a = [1, 2, [3, 4]]

b = copy.copy(a)       # 浅拷贝
b[2].append(5)
print(a)               # [1, 2, [3, 4, 5]]，原对象被影响

c = copy.deepcopy(a)   # 深拷贝
c[2].append(6)
print(a)               # [1, 2, [3, 4, 5]]，原对象不受影响
```

````python
import copy

nums = [[1, 2, 3], 4, 5]
nums_copy = copy.copy(nums)
nums_deepcopy = copy.deepcopy(nums)
nums_copy[0][0] = 1000
nums_copy[2] = 10
nums, nums_copy, nums_deepcopy
````

## 57. Python 中的列表和元组有什么区别?
二者都是有序序列容器，核心差异如下：

| 特性 | 列表（list） | 元组（tuple） |
|------|--------------|--------------|
| 可变性 | 可变，支持增删改 | 不可变，创建后无法修改元素 |
| 语法 | 方括号 `[]` 定义 | 圆括号 `()` 定义，单元素需加逗号 `(1,)` |
| 内置方法 | 方法丰富（append、pop、sort 等） | 方法极少，仅 count、index 等 |
| 内存占用 | 动态扩容，占用更高 | 固定大小，占用更小，性能更优 |
| 哈希性 | 不可哈希，不能做字典键 | 元素均不可变时可哈希，可做字典键 |
| 适用场景 | 存储需动态修改的数据 | 存储固定数据、函数多返回值、保护数据不被篡改 |

> 补充：元组的不可变指元素引用不可变；若元组内包含列表等可变对象，其内部内容仍可修改。

````python
# --- 可变性
nums1 = [1, 2, 3]
nums2 = (1, 2, 3)
nums1[1] = -1
# nums1
# nums2[1]=-2
# 内置方法
nums1.index(1)
# --- 可哈希
dicts={
    nums2:1,
    # nums1:2,
}
dicts
````

## 58. 什么是 Python 的字典，有哪些用法?
字典（dict）是基于哈希表实现的键值对容器，Python 3.7+ 为插入有序。键必须是不可变、可哈希的类型，值可以是任意对象。

核心用法：
1. **创建字典**
```python
dict1 = {'name': 'Tom', 'age': 20}
dict2 = dict(name='Tom', age=20)
dict3 = dict([('name', 'Tom'), ('age', 20)])
```
2. **访问元素**：`dict['key']` 键不存在报错；`get(key, default)` 更安全，不存在返回默认值。
3. **增删改**
```python
dict1['gender'] = '男'   # 新增/修改
del dict1['age']         # 删除指定键
dict1.pop('name')        # 删除并返回值
dict1.clear()            # 清空字典
```
4. **遍历**
```python
for k in dict1: pass             # 遍历键
for v in dict1.values(): pass    # 遍历值
for k, v in dict1.items(): pass  # 遍历键值对
```
5. 其他：`update()` 合并字典、`setdefault()` 设置默认值、`keys()` 获取所有键等。

````python
dicts = {
    "jom": 1,
    "kom": 2,
    "lom": 3,
}
print(dicts)
print(dicts.items())
index = 1
for k, v in dicts.items():
    print(f"[{index}] >>> {k},{v}")
    index += 1
````

## 60. 什么是 Python 的闭包?
闭包指：内部函数引用了外部函数作用域的变量，且外部函数返回该内部函数，这个内部函数就是闭包。

闭包三要素：
1. 函数嵌套（内部函数定义在外部函数内）
2. 内部函数引用外部函数的变量
3. 外部函数返回内部函数

作用：保存外部函数的变量状态，延长变量生命周期，即使外部函数执行完毕，内部函数仍可访问外部变量。

```python
def outer(x):
    def inner(y):
        return x + y  # 引用外部变量 x
    return inner

add5 = outer(5)
print(add5(3))   # 8
print(add5(10))  # 15，x 的值被保留
```
> 注意：闭包内修改外部变量需用 `nonlocal` 关键字声明。

````python
def outer(x):
    def inner(y):
        return x + y  # 引用外部变量 x

    return inner


add5 = outer(5)
print(type(add5))
print(add5(3))  # 8
print(add5(10))  # 15，x 的值被保留
````
